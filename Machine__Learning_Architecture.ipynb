```python
# import Necessary Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings(action = 'ignore')
import logging
logging.basicConfig(level=logging.INFO,
                    filename = 'model.log',
                    filemode = 'w',
                    format = '%(asctime)s - %(message)s - %(levelname)s',
                    force = 'True')
```

```python
# Step 1 : Data Ingestion

def data_ingestion():
  '''
  *Title: Composite Cement Strength Prediction Model

  *Shape of the Dataset: (1030, 10)

  *Dataset: Concrete_Data.csv

  *Independent Columns: ['Cement (component 1)(kg in a m^3 mixture)',
       'Blast Furnace Slag (component 2)(kg in a m^3 mixture)',
       'Fly Ash (component 3)(kg in a m^3 mixture)',
       'Water  (component 4)(kg in a m^3 mixture)',
       'Superplasticizer (component 5)(kg in a m^3 mixture)',
       'Coarse Aggregate  (component 6)(kg in a m^3 mixture)',
       'Fine Aggregate (component 7)(kg in a m^3 mixture)', 'Age (day)',
       'Water_to_Cementitious_MaterialsRation '],

  *Target Columns: ['Concrete compressive strength(MPa, megapascals) ']
  '''
  DF = pd.read_csv('/content/Concrete_Data - Concrete_Data.csv')
  return DF
```

```python
DF = data_ingestion()
```

```python
# Step 2 : Data Reprocessing and Cleaning

from sklearn.preprocessing import StandardScaler, MinMaxScaler, RobustScaler
from sklearn.model_selection import train_test_split

def data_preprocessing(DF):
  #Step 1: Feature Engineering and Feature Selection
  DF['Water_to_Cementitious_MaterialsRatio'] = DF['Water  (component 4)(kg in a m^3 mixture)'] / (
      DF['Cement (component 1)(kg in a m^3 mixture)'] +
      DF['Blast Furnace Slag (component 2)(kg in a m^3 mixture)'] +
      DF['Fly Ash (component 3)(kg in a m^3 mixture)']
  )

  #Step2 : Split the dataset into X and y
  target_col = 'Concrete compressive strength(MPa, megapascals) '
  X = DF.drop(columns = [target_col])
  y = DF[target_col]

  #Step3 :- Split the Dataset into Seen and Unseen (Seen Data: train data and Unseen data: test Data)
  X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1)

  #Step4 : Normalization / Scaling
  sc = MinMaxScaler()
  X_train = sc.fit_transform(X_train)
  X_test = sc.transform(X_test)

  return X_train, X_test, y_train, y_test
```

```python
# Step 3 : Model Training and Evaluation

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

def model_build(X_train,X_test,y_train,y_test):
  # Define Model
  LR=LinearRegression()
  LR.fit(X_train,y_train) # Fitting it on Seen Data
  y_pred=LR.predict(X_test) # Prediction is on Unseen Data

  r2=r2_score(y_test,y_pred) # (Actual Value, Predicted Value)

  #Checking Model Performance Plot1
  sns.displot(y_test - y_pred)
  plt.show()

  #Checking Model Performance Plot1
  sns.regplot(x=y_test,y=y_pred)
  plt.show()
  return r2
```

```python
#Orchestron Layer : Entry Point

def main(): #Function Definition
  #Step1: Data Ingestion
  DF = data_ingestion()

  # Step 2: Data Preprocessing and Cleaning
  X_train,X_test,y_train,y_test = data_preprocessing(DF)

  #Step 3: Model Building
  r2 = model_build(X_train,X_test,y_train,y_test)
  print(f'Linear Regression R2 score {r2*100} is')
```
